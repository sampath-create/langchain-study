### Model integration with Gemini,OpenAi,Groq

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

### Gemini

```python
import os
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
```

```python
response = model.invoke("Explain about Himalayas").content
response
```

### Groq

```python
import os
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("Explain about Himalayas").content
response
```

### streaming
The output is displayed as soon it is generated using stream

```python
for chunk in model.stream("Explain about AIin 100 words"):
    print(chunk.text,end = "",flush=True)
```

### Batch
Use it for pareel asking of questions

```python
responses=model.batch({
    "who is Micheal jackson",
    "what is Machine Learning",
    "where is Andhra Pradesh Located"
})
for i in responses:
    print(i)
```
