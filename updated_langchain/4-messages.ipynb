## Messages
LangChain uses dedicated message classes to represent different roles in a conversation, ensuring a unified schema across various LLM providers (e.g., OpenAI, Anthropic, Gemini). The core classes include:

### SystemMessage: 
Used to set the LLM's behavior, instructions, or persona.
### HumanMessage: 
Represents messages sent by the user.
### AIMessage: 
Represents responses generated by the model (which can also carry metadata like tool_calls).
### ToolMessage / FunctionMessage:
 Used to pass the execution results of tools/functions back to the model.


```python
import os
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("Explain about Tumor").content
response
```

```python
#Text Prompts:- Stand alone request and maintain no conversation History
model.invoke("What is langchain")
```

```python
from langchain.messages import HumanMessage,AIMessage,SystemMessage
messages=[
    SystemMessage("Your are an coding Assistent"),
    HumanMessage("Explain abour REST API")
]
response=model.invoke(messages).content
print(response)
```

```python
# detailed system message for better perrformance

messages=[
    SystemMessage("Your are an coding Assistent,who has 12 year experience in SOftware always provide with examples and do reasoning on it , be consise throughout your explanations"),
    HumanMessage("Explain abour REST API")
]
response=model.invoke(messages).content
print(response)
```

```python
#Message Meta Data
human_message=HumanMessage(
    content = "Hi how are you gemini!",
    name="Ruku",
    id="1433"
)
response=model.invoke(([human_message])).content
print(response)
```

```python
#Ai message
ai_mesage=AIMessage("i'm happy to help you with thw question")
messages=[
        SystemMessage("Your are an coding Assistent,who has 12 year experience in SOftware always provide with examples and do reasoning on it , be consise throughout your explanations"),
        HumanMessage("CAn you Help Me"),
        ai_mesage,
        HumanMessage("Can you explain about RAG")
]
response=model.invoke(messages)
print(response.content)
```

```python
from langchain.messages import ToolMessage

ai_message=AIMessage(
    content=[],
    tool_calls=[{
        "name":"get_weather",
        "args":{"location":"Guntur VVIT"},
        "id":"1432"
    }]
)
weather_result="sunny, 44 degrees"
toll_message=ToolMessage(
    content=weather_result,
    tool_call_id="1432"
)
messages=[
    HumanMessage("What is the weather in Guntur VVIT "),
    ai_message,
    toll_message
]
response=model.invoke(messages)
```
