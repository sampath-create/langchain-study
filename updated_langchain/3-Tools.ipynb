### Tools

```python
import os
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("Explain about Himalayas").content
response
```

```python
from langchain.tools import tool
@tool
def get_weather(location:str)->str:
    """ Get wather in a location """
    return f"weather in {location} is sunny"
model_tools=model.bind_tools([get_weather])
```

```python
response=model_tools.invoke("whats the weather in Mumbai")
for i in response.tool_calls:
    print(i["name"])
    print(i["args"])
```
